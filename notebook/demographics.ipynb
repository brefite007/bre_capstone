```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
location = pd.read_csv('../data/demo_location.csv')
location
```

```python
avg_location = location.groupby(['Location', 'Food Type'])['Total'].mean().reset_index()
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=avg_location, x='Location', y='Total', hue='Food Type')
plt.title('Average Annual Food & Beverage Expenditure by Region (2019-2023)', fontsize=14)
plt.ylabel('Avg Annual Spending (Millions)')
plt.xlabel('Region')
plt.legend(title='Food & Beverage Type')
plt.show()
```

```python
income = pd.read_csv('../data/demo_income.csv')
income.head()
```

```python
income_order = ['0-14', '15-29', '30-39', '40-49', '50-69', '70-99', '100-149', '150-199', '200+']
```

```python
relevant_types = ['Food At Home', 'Food Away From Home', 'Alcoholic Beverages']
subset = income[income['Type'].isin(relevant_types)]
```

```python
avg_spending = subset.groupby(['Income', 'Type'])['Total'].mean().reset_index()
```

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=avg_spending, x='Income', y='Total', hue='Type')

plt.title('Average Annual Spending on Food and Alcohol by Income')
plt.xlabel('Income Before Taxes ($1000s)')
plt.ylabel('Avg Spending (Millions)')
plt.xticks(rotation=45)
plt.legend(title='Spending Category')
plt.savefig('avg_ann_spend_food_alc.png', dpi=300)
plt.show()
```
