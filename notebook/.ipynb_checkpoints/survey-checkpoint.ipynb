```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dining = pd.read_csv('../data/dining_takeout_delivery.csv')
dining
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=dining, x='Response Detail', y='Value', hue='Location')
for container in plt.gca().containers:
    plt.bar_label(container, fmt='%.0f%%', label_type='edge', padding=3)
plt.title('Dining and Takeout Frequency: USA vs. WV')
plt.ylabel('Times per Month')
plt.tight_layout()
plt.savefig('dining_out.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
preference = pd.read_csv('../data/consumer_preference_survey.csv')
preference
```

```python
preference.columns = preference.columns.str.strip()
preference['Value'] = preference['Value'].str.replace('%', '').astype(float)
```

```python
dining_out = preference[preference['Response Detail'] == 'reasons for dining out']
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=dining_out, y='Reason', x='Value', hue='Location')
plt.title('Top Reasons for Dining Out: USA vs. WV')
plt.xlabel('Percentage of Responses')
plt.tight_layout()
plt.savefig('dining_out.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
takeout = preference[preference['Response Detail'] == 'reasons for ordering takeout or delivery']
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=takeout, y='Reason', x='Value', hue='Location')
plt.title('Top Reasons for Ordering Takeout/Delivery: USA vs. WV')
plt.xlabel('Percentage of Responses')
plt.tight_layout()
plt.savefig('takeout_delivery.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
spending = preference[preference['Response Detail'] == 'dining out spending']

plt.figure(figsize=(8, 5))
sns.barplot(data=spending, x='Reason', y='Value', hue='Location')

for container in plt.gca().containers:
    plt.bar_label(container, fmt='%.0f%%', label_type='edge', padding=2)

plt.title('Dining Out Spending Distribution: USA vs. WV')
plt.xlabel('Spending Bracket (Per Person)')
plt.ylabel('Percentage of Responses')
plt.tight_layout()
plt.savefig('spending_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
dine_in = preference[preference['Response Detail'] == 'restaurant type dining']

plt.figure(figsize=(8, 5))
sns.barplot(data=dine_in, x='Reason', y='Value', hue='Location')

for container in plt.gca().containers:
    plt.bar_label(container, fmt='%.0f%%', label_type='edge', padding=2)

plt.title('Preferred Restaurant Type – Dining In')
plt.xlabel('Restaurant Type')
plt.ylabel('Percentage of Responses')
plt.tight_layout()
plt.savefig('restaurant_type_dining.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python

takeout = preference[preference['Response Detail'] == 'restaurant type for takeout and delivery ']


plt.figure(figsize=(8, 5))
sns.barplot(data=takeout, x='Reason', y='Value', hue='Location')


for container in plt.gca().containers:
    plt.bar_label(container, fmt='%.0f%%', label_type='edge', padding=2)


plt.title('Preferred Restaurant Type – Takeout & Delivery')
plt.xlabel('Restaurant Type')
plt.ylabel('Percentage of Responses')
plt.tight_layout()
plt.savefig('restaurant_type_takeout_delivery.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python

tod_pref = preference[preference['Response Detail'] == 'preference for takeout or delivery']


plt.figure(figsize=(8, 5))
sns.barplot(data=tod_pref, x='Reason', y='Value', hue='Location')


for container in plt.gca().containers:
    plt.bar_label(container, fmt='%.0f%%', label_type='edge', padding=3)

plt.title('Takeout vs Delivery Preference – USA vs West Virginia')
plt.xlabel('Preference')
plt.ylabel('Percentage of Responses')
plt.tight_layout()
plt.savefig('takeout_vs_delivery.png', dpi=300, bbox_inches='tight')
plt.show()
```
